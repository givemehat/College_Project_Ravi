# ***Find Meta Data***

```python
import glob
import os

metadata_files = glob.glob(
    "/content/battery_dataset/**/metadata.csv",
    recursive=True
)

if not metadata_files:
    raise FileNotFoundError("metadata.csv not found inside the dataset.")

METADATA_PATH = metadata_files[0]

DATA_DIR = os.path.join(
    os.path.dirname(METADATA_PATH),
    "data"
)

print("Metadata:", METADATA_PATH)
print("Data directory:", DATA_DIR)
```

# **Cell 3 Load & Clean Data**

```python
df = pd.read_csv(METADATA_PATH)
df["Capacity"] = pd.to_numeric(
    df["Capacity"]
    .astype(str)
    .str.replace(r"[\[\]]", "", regex=True),
    errors="coerce"
)
discharge = (
    df[
        (df["type"] == "discharge") &
        (df["Capacity"] > 0.05)
    ]
    .sort_values(["battery_id", "test_id"])
    .copy()
)
print("Total discharge records:", len(discharge))
print("Total batteries:", discharge["battery_id"].nunique())
```

# **Cell 4 Calculate Initial Capacity & SOH**

```python
# Robust formation capacity (filtering degenerate cycles < 1.0 Ah)
valid_dis = discharge[discharge["Capacity"] > 1.0].copy()
initial_capacity = valid_dis.groupby("battery_id")["Capacity"].apply(
    lambda s: s.iloc[:3].mean() if len(s) >= 3 else s.max()
)

discharge["Initial_Capacity"] = discharge["battery_id"].map(initial_capacity)
discharge["cycle_index"] = discharge.groupby("battery_id").cumcount() + 1
discharge["SOH"] = np.clip(discharge["Capacity"] / discharge["Initial_Capacity"] * 100, 0.0, 100.0)

discharge[["battery_id", "cycle_index", "test_id", "Capacity", "SOH"]].head()

```

# **Cell 5 Calculate RUL**

```python
def calculate_rul(group):
    eol_threshold = 0.70 * group["Initial_Capacity"].iloc[0]
    eol_cycles = group.loc[group["Capacity"] <= eol_threshold, "cycle_index"]
    eol_cycle = eol_cycles.iloc[0] if not eol_cycles.empty else group["cycle_index"].max()
    group["EOL_Cycle"] = eol_cycle
    group["RUL"] = np.maximum(0, eol_cycle - group["cycle_index"])
    return group

discharge = discharge.groupby("battery_id", group_keys=False).apply(calculate_rul)
discharge[["battery_id", "cycle_index", "test_id", "SOH", "RUL"]].head()

```

# **Cell 6 Capacity Degradation Plot**

```python
# Battery Capacity Degradation FacetGrid across Operating Temperatures with 2-Column Legend
discharge['Cell_Type'] = 'Operating Temp: ' + discharge['ambient_temperature'].astype(str) + '°C'

g = sns.FacetGrid(
    discharge[discharge['Capacity'] > 1.0],
    col='Cell_Type',
    col_wrap=3,
    height=4.6,
    aspect=1.1,
    sharex=False,
    sharey=True
)
g.map_dataframe(
    sns.lineplot,
    x='cycle_index',
    y='Capacity',
    hue='battery_id',
    palette='tab20',
    alpha=0.85,
    lw=1.5
)
g.add_legend()
sns.move_legend(
    g,
    loc='center left',
    bbox_to_anchor=(1.01, 0.5),
    ncol=2,
    title='Battery Cell ID',
    frameon=True,
    fontsize=8,
    title_fontsize=9.5
)
g.set_axis_labels('Discharge Cycle Index', 'Discharge Capacity (Ah)', fontsize=11, fontweight='medium')
g.set_titles(col_template='{col_name}', size=12, weight='bold')
plt.subplots_adjust(top=0.86, right=0.80)
g.fig.suptitle('Battery Discharge Capacity Degradation across Operating Temperatures', fontsize=14, fontweight='bold', y=0.97)
plt.show()
```

# **Cell 7 SOH Degradation Plot**

```python
# Regenerated Discharge Capacity vs Cycle Index and SOH Degradation Trajectories
fig, axes = plt.subplots(1, 2, figsize=(16, 5.5))

# Left: Discharge Capacity vs Cycle Index
sns.lineplot(
    data=discharge[discharge['Capacity'] > 1.0],
    x='cycle_index',
    y='Capacity',
    hue='battery_id',
    palette='tab20',
    alpha=0.65,
    lw=1.3,
    legend=False,
    ax=axes[0]
)
axes[0].axhline(1.4, color='#d95f02', linestyle='--', lw=2.0, label='Nominal 70-80% EOL Zone (~1.4 Ah)')
axes[0].axhline(2.0, color='#2ca02c', linestyle=':', lw=1.8, label='Rated Fresh Capacity (~2.0 Ah)')
axes[0].set_title('Discharge Capacity vs Cycle Index (All Cells)', fontsize=12.5, fontweight='bold')
axes[0].set_xlabel('Discharge Cycle Index', fontsize=11)
axes[0].set_ylabel('Discharge Capacity (Ah)', fontsize=11)
axes[0].set_ylim(0.9, 2.2)
axes[0].legend(frameon=True, loc='upper right')

# Right: SOH Degradation Trajectories
sns.lineplot(
    data=discharge[discharge['Capacity'] > 1.0],
    x='cycle_index',
    y='SOH',
    hue='battery_id',
    palette='tab20',
    alpha=0.60,
    lw=1.3,
    legend=False,
    ax=axes[1]
)
axes[1].axhline(80, linestyle='--', color='#e6ab02', lw=2.0, label='80% SOH Warning Limit')
axes[1].axhline(70, linestyle='--', color='#d95f02', lw=2.0, label='70% SOH EOL Boundary')
axes[1].set_title('SOH Degradation Trajectories vs Cycle Index', fontsize=12.5, fontweight='bold')
axes[1].set_xlabel('Discharge Cycle Index', fontsize=11)
axes[1].set_ylabel('State of Health SOH (%)', fontsize=11)
axes[1].set_ylim(45, 105)
axes[1].legend(frameon=True, loc='upper right')

plt.tight_layout()
plt.show()
```

# **Cell 8 RUL Distribution**

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# Left: RUL Degradation Trajectories (Linear countdown to EOL)
sns.lineplot(
    data=discharge[discharge["Capacity"] > 1.0],
    x="cycle_index",
    y="RUL",
    hue="battery_id",
    ax=axes[0],
    alpha=0.7
)
axes[0].axhline(0, color='red', linestyle='--', label='EOL Boundary (RUL = 0)')
axes[0].set_title("Remaining Useful Life (RUL) Trajectory Degradation Curves", fontweight='bold')
axes[0].set_xlabel("Discharge Cycle Number (Consecutive Cycles)")
axes[0].set_ylabel("RUL (Cycles Remaining)")
axes[0].legend()
axes[0].grid(True, linestyle=':', alpha=0.6)

# Right: RUL Distribution Histogram
sns.histplot(
    discharge[discharge["Capacity"] > 1.0]["RUL"],
    bins=30,
    kde=True,
    color='#1f77b4',
    ax=axes[1]
)
axes[1].set_title("Remaining Useful Life (RUL) Distribution Across Operating Cycles", fontweight='bold')
axes[1].set_xlabel("RUL (Cycles Remaining)")
axes[1].set_ylabel("Number of Samples")
axes[1].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()

```

# **Cell 9 Battery Quality Audit**

```python
discharge["Capacity_Change"] = (
    discharge
    .groupby("battery_id")["Capacity"]
    .diff()
    .abs()
)

audit = (
    discharge
    .groupby("battery_id")
    .agg(
        Total_Cycles=("test_id", "count"),
        Min_Capacity=("Capacity", "min"),
        Max_Capacity=("Capacity", "max"),
        Mean_Capacity_Change=("Capacity_Change", "mean")
    )
    .reset_index()
)

display(audit)
```

# **Cell 10 Identify Difficult and Anomalous Batteries**

```python
difficult_cells = audit[
    (audit["Mean_Capacity_Change"] > 0.04) |
    (audit["Total_Cycles"] < 30)
]

print("Potentially Difficult / Anomalous Batteries:")

display(difficult_cells)
```
